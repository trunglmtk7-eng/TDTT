W4A18 Nhập vào số nguyên dương a và b, in toàn bộ ước chung của a và b

```python
a, b = map(int, input().split())
if a>b:
    MIN=a
else:
    MIN=b
for i in range(1, MIN+1):
    if a%i==0 and b%i==0:
        print(i , end=" ")
```

W4A19 Viết chương trình Python sử dụng một trong 2 loại vòng lặp for hoặc while để in ra các số 
dương chẵn nhỏ hơn hoặc bằng n (n <= 1000)

```python
n = int(input("Nhập n (n <= 1000): "))
i = 2 
while i <= n:
    print(i, end=" ")
    i += 2
```

W4A20 Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 2^k hay không

```python
def check_2k(n):
    if n <= 0:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1
n = int(input("Nhập n: "))
if check_2k(n):
    print(f"{n} là số dạng 2^k")
else:
    print(f"{n} không phải là số dạng 2^k")
```

W4A21 Nhập vào số nguyên dương n, tính tổng các chữ số của n

```python
n = int(input())
tong=0
while n>0:
    so_cuoi = n%10
    tong += so_cuoi
    n//=10
print(tong)

```

W4A22 Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ

```python
n = int(input())
le=0
chan=0
while n>0:
    so_cuoi = n%10
    if so_cuoi%2==0:
        chan +=1
    else:
        le +=1
    n//=10
print(f"so le: {le} va so chan: {chan}")
```

W4A23 Nhập n 
Cho S(k) = 1 + 2 + 3 + … + k 
Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n

```python
n = int(input())
k=0
tong = 0
while tong + (k+1)<n:
    k+=1
    tong+=k
print(f" gia tri k la {k}")
```

W4A24 Nhập vào A, tìm n nhỏ nhất sao cho 
1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A

```python
A = float(input())
n=0
tong=0
while tong<=A:
    n +=1
    tong +=1/n
print(f"gia tri n la {n}")
```
