Bài 6. Nhập vào 3 số a, b, c. Kiểm tra xem đây có phải là cạnh của 1 tam giác không, 
nếu đúng thì tính diện tích của tam giác đó, in ra với 1 chữ số phần thập phân, nếu nó 
không phải 3 cạnh của tam giác thì in ra “Khong phai 3 canh cua tam giac”.

```python
a, b, c= map(float, input().split())
if a+b>c and a+c>b and b+c>a:
    p=(a+b+c)/2
    print(f"{(p*(p-a)*(p-b)*(p-c))**0.5:.1f}")
else:
    print("Khong phai 3 canh cua tam giac")
```

Kiểm tra ký tự nhập vào là chữ hay số: Viết chương trình nhập một ký tự, kiểm 
tra xem ký tự đó là chữ cái hay số.

```python
c = str(input("Nhập vào một ký tự: "))
if len(c) == 1:
    if c.isalpha():
        print(f"{c} là chữ cái.")
    elif c.isdigit():
        print(f"{c} là chữ số.")
    else:
        print(f"{c} không phải là chữ cái cũng không phải là chữ số (có thể là ký tự đặc biệt).")
else:
    print("Vui lòng chỉ nhập đúng một ký tự.")
```
