Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểm tra xem nó có thể là
độ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh 
cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, 
Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra màn
hình

```python
a, b, c= map(int, input().split())
if a+b>c and a+c>b and b+c>a:
    if a==b==c:
        print("tam giac deu")
    elif a==b or a==c or b==c:
        print("tam giac can")
    elif (a**2 + b**2 == c**2) or (a**2 + c**2 == b**2) or (b**2 + c**2 == a**2):
        print("tam giac vuong")
    else:
        print("tam giac thuong")
else:
    print("Khong la ba canh cua mot tam giac")
```

Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
a, b, c=map(float, input().split())
if a == 0:
    if b == 0:
        if c == 0:
            print("Phuong trinh vo so nghiem")
        else:
            print("Phuong trinh vo nghiem")
    else:
        print(f"Phuong trinh co nghiem duy nhat: x = {-c/b:.2f}")
else:
    d = b**2 - 4*a*c
    if d > 0:
        x1 = (-b + d**0.5) / (2*a)
        x2 = (-b - d**0.5) / (2*a)
        print(f"x1 = {x1:.2f}")
        print(f"x2 = {x2:.2f}")
    elif d == 0:
        x = -b / (2*a)
        print(f"Phuong trinh co nghiem kep: x1 = x2 = {x:.2f}")
    else:
        print("Phuong trinh co nghiem phuc")
```

Bài 8: Giải hệ phương trình tuyến tính:
ax + by= m
cx + dy = n

```python
a, b, c, d, m, n = map(int, input().split())
k = a * d - b * c
kx = m * d - b * n
ky = a * n - m * c
if k!= 0:
    x = kx / k
    y = ky / k
    print(f"Hệ phương trình có nghiệm duy nhất: x = {x:.2f}, y = {y:.2f}")
else:
    if kx == 0 and ky == 0:
        print("Hệ phương trình có vô số nghiệm")
    else:
        print("Hệ phương trình vô nghiệm")
```

Bài 9: Nhập vào thời gian 1 công việc nào đó là x giây. Hãy chuyển đổi và viết 
ra màn hình số thời gian trên dưới dạng bằng bao nhiêu giờ, bao nhiêu phút, bao 
nhiêu giây

```python
x = int(input("Nhập vào số giây (x): "))
gio = x // 3600
phut = (x % 3600) // 60
giay = x % 60
print(gio, phut, giay)
```
